```python
# prompt
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
You are an expert learning path designer.

Create a practical and realistic learning roadmap
based on the user's requirements.

The roadmap must:
1. Start from the user's current level.
2. Progress from fundamentals to advanced concepts.
3. Include practical projects.
4. Respect the available duration.
5. Match the user's goal.
6. Avoid unnecessary topics.
"""
    ),
    (
        "human",
        """
Create a learning path for:

Topic: {topic}

Current Level: {level}

Goal: {goal}

Duration: {duration}

Hours Available Per Week: {hours_per_week}
"""
    )
])

print("Prompt created successfully!")
```

```python
from IPython.display import display, Markdown

topic = "Generative AI"
level = "Beginner"
goal = "Become job ready"
duration = "16 weeks"
hours_per_week = 10

formatted_prompt = prompt.invoke({
    "topic": topic,
    "level": level,
    "goal": goal,
    "duration": duration,
    "hours_per_week": hours_per_week
})

display(Markdown(f"""
# 🚀 Learning Path Generator

### 📌 Your Requirements

| Requirement | Details |
|---|---|
| 📚 **Topic** | {topic} |
| 🎯 **Current Level** | {level} |
| 💼 **Goal** | {goal} |
| ⏳ **Duration** | {duration} |
| ⏰ **Hours / Week** | {hours_per_week} hours |

---

## 🧠 Prompt Structure

### 🤖 System Instructions

> You are an expert learning path designer.

The AI is instructed to:
- Start from the user's current level
- Progress from fundamentals to advanced concepts
- Include practical projects
- Respect the available duration
- Match the user's goal
- Avoid unnecessary topics

### 👤 User Request

**Create a learning path for {topic}**

The user's requirements are:

**Level:** {level}
**Goal:** {goal}
**Duration:** {duration}
**Hours per week:** {hours_per_week}

---

✅ **Prompt successfully prepared for Gemini**
"""))
```

```python
from pydantic import BaseModel, Field
from typing import List


class LearningStage(BaseModel):
    stage: str = Field(
        description="Name of the learning stage"
    )

    duration: str = Field(
        description="Duration of this stage"
    )

    topics: List[str] = Field(
        description="Important topics to learn in this stage"
    )

    project: str = Field(
        description="Practical project for this stage"
    )


class LearningPath(BaseModel):
    title: str = Field(
        description="Title of the learning roadmap"
    )

    overview: str = Field(
        description="Short overview of the complete roadmap"
    )

    prerequisites: List[str] = Field(
        description="Prerequisites required before starting"
    )

    stages: List[LearningStage] = Field(
        description="Ordered learning stages"
    )

    final_project: str = Field(
        description="Final project recommendation"
    )


print("Pydantic models created successfully! ✅")
```

```python
structured_llm = llm.with_structured_output(LearningPath)

print("Structured Gemini model created successfully! ✅")
```

```python
result = structured_llm.invoke(
    """
    Create a learning path for:

    Topic: Generative AI
    Current Level: Beginner
    Goal: Become job ready
    Duration: 16 weeks
    Hours Available Per Week: 10
    """
)

print(result)
```

```python
from IPython.display import display, Markdown

md = f"""
# 🚀 {result.title}

## 📝 Overview

{result.overview}

## 📋 Prerequisites

"""

for prerequisite in result.prerequisites:
    md += f"- ✅ {prerequisite}\n"

md += "\n## 📚 Learning Stages\n\n"

for stage in result.stages:
    md += f"""
### 🎯 {stage.stage}

**⏳ Duration:** {stage.duration}

**📖 Topics:**

"""

    for topic in stage.topics:
        md += f"- {topic}\n"

    md += f"""

**🛠️ Practical Project:**
{stage.project}

---

"""

md += f"""
## 🏆 Final Project

{result.final_project}
"""

display(Markdown(md))
```

```python
chain = prompt | structured_llm
```

```python
def generate_learning_path(
    topic,
    level,
    goal,
    duration,
    hours_per_week
):

    result = chain.invoke({
        "topic": topic,
        "level": level,
        "goal": goal,
        "duration": duration,
        "hours_per_week": hours_per_week
    })

    return result


print("Learning Path Generator function created! ✅")
```

```python
result = generate_learning_path(
    topic="Python",
    level="Beginner",
    goal="Become job ready",
    duration="8 weeks",
    hours_per_week=10
)

print(result.title)
```

```python
print("TITLE:")
print(result.title)

print("\nOVERVIEW:")
print(result.overview)

print("\nPREREQUISITES:")
for item in result.prerequisites:
    print("-", item)

print("\nSTAGES:")
for stage in result.stages:
    print(f"\n{stage.stage}")
    print("Duration:", stage.duration)
    print("Topics:", stage.topics)
    print("Project:", stage.project)

print("\nFINAL PROJECT:")
print(result.final_project)
```
